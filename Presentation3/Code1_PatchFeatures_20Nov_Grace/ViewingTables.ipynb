# Data Exploration

### Patches

```python
# High-grade, low-grade, and benign counts
PATCHES_DIR = "/projects/e32998/patch_images"
png_ids = {os.path.splitext(f)[0]
           for f in os.listdir(PATCHES_DIR)
           if f.lower().endswith(".png")}

parquet_ids = set(patches["patch_full"].astype(str))
missing = png_ids - parquet_ids

print("PNG files:", len(png_ids))
print("Parquet patches:", len(parquet_ids))
print("Missing from Parquet:", len(missing))

missing_df = pd.DataFrame({"patch_id": sorted(missing)})
missing_df["case"] = (missing_df["patch_id"].str.extract(r"case_(\d+)")[0].astype(int))
display(missing_df.groupby("case").size())
print(patches["label"].value_counts())
```

```python
# Embeddings
missing = patches[patches["emb_path"].isna()]
print("Missing embeddings: ", missing.shape[0])
print(((missing["width"] < 32) | (missing["height"] < 32)).all())
print("New cases, no embedding yet: ", (patches["emb_path"] == "N/A (new case)").sum())
print("Embedding available: ", (patches["emb_path"].notna() & (patches["emb_path"] != "N/A (new case)")).sum())
```

```python
# Patch breakdown by stain
print(patches["stain"].value_counts()) # counts
print(patches["stain"].value_counts(normalize=True)) # proportions
```

```python
# Patch dimensions
patches[["width", "height", "patch_size"]].describe()
```
