```python
import os
import pandas as pd

PATCHES_DIR = "/projects/e32998/patch_images"

png_ids = {
    os.path.splitext(f)[0]
    for f in os.listdir(PATCHES_DIR)
    if f.lower().endswith(".png")
}

parquet_ids = set(patches["patch_id"].astype(str))

missing = png_ids - parquet_ids

print("PNG files:", len(png_ids))
print("Parquet patches:", len(parquet_ids))
print("Missing from Parquet:", len(missing))

# Show missing patches
missing_df = pd.DataFrame({"patch_id": sorted(missing)})
missing_df["case"] = (
    missing_df["patch_id"]
    .str.extract(r"case_(\d+)")[0]
    .astype(int)
)

display(missing_df.groupby("case").size())
```
