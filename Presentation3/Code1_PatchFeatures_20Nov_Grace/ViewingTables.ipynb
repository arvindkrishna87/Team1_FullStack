### Flagging Unsual Cases

```python
# Features to examine (can change)
features = [#"n_patches",
            #"n_slices",
            "mean_patch_size",
            "patch_size_iqr",
            "patches_per_slice",
            "small_patch_share",
            "h_e_share",
            "melan_share",
            "sox10_share"]
```

```python
# Store unsual cases
flagged = []
for feature in features:
    lower = cases[feature].quantile(0.25) - 1.5 * (cases[feature].quantile(0.75) - cases[feature].quantile(0.25))
    upper = cases[feature].quantile(0.75) + 1.5 * (cases[feature].quantile(0.75) - cases[feature].quantile(0.25))
    outliers = cases[(cases[feature] < lower) | (cases[feature] > upper)]
    for _, row in outliers.iterrows():
        flagged.append({"case_id": row["case_id"],
                        "grade": row["grade"],
                        "feature": feature,
                        "value": row[feature],
                        "direction": ("High" if row[feature] > upper else "Low")})
flagged_df = pd.DataFrame(flagged, columns=["case_id", "grade", "feature", "value", "direction"])
flagged_df.head(5)
```

```python
# Get top unusual cases
case_flags = (flagged_df.groupby("case_id").agg(n_unusual_features=("feature", "count"),
                                                unusual_features=("feature", lambda x: ", ".join(x))).reset_index().sort_values("n_unusual_features", ascending=False))
case_flags = case_flags.merge(cases[["case_id", "grade", "n_patches"]], on="case_id", how="left")
case_flags.head(10)
```

```python
# Print unusual values
for case_id in [16, 17, 38, 46]:
    print(f"\nCase {case_id}:")
    case_info = cases[cases["case_id"] == case_id].iloc[0]
    print(f"Patient: {case_info['patient_id']}")
    print(f"Label: {case_info['label']}")
    print(f"Grade: {case_info['grade']}")
    print(f"Number of patches: {case_info['n_patches']}")
    print(f"Number of slices: {case_info['n_slices']}")
    print(f"Number of stains: {case_info['n_stains']}")
    print(f"Stains: {case_info['stains']}")
    print(f"Missing embeddings: {case_info['n_missing_emb']}")
    print(flagged_df[flagged_df["case_id"] == case_id][["feature", "value", "direction"]].to_string(index=False))
```
